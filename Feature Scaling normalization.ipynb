```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df  = pd.read_csv('Wine dataset.csv' , usecols=[0,1,2])
df.columns=['Class label','Alcohal','Malic acid']
```

```python
df
```

```python
sns.kdeplot(df['Alcohal'])
```

```python
sns.kdeplot(df['Malic acid'])
```

```python
color_dict = {1: 'red', 2: 'blue', 3: 'green'}

sns.scatterplot(
    data=df,
    x='Alcohal',
    y='Malic acid',
    hue='Class label',
    palette=color_dict
)

plt.show()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    df.drop('Class label', axis=1),
    df['Class label'],
    test_size=0.3,
    random_state=0
)
```

```python
x_train.shape, x_test.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(x_train)

x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
scaler.mean_
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled,columns=x_train.columns)
x_test_scaled = pd.DataFrame(x_train_scaled,columns=x_test.columns)
```

```python
np.round(x_train.describe(),1)
```

```python
np.round(x_train_scaled.describe(),1)
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# Before Scaling
ax1.scatter(x_train['Alcohal'], x_train['Malic acid'])
ax1.set_title('Before Scaling')


# After Scaling
ax2.scatter(x_train_scaled['Alcohal'], x_train_scaled['Malic acid'])
ax2.set_title('After Scaling')


plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# Before Scaling
ax1.set_title('Before Scaling')
sns.kdeplot(x_train['Alcohal'],ax=ax1)
sns.kdeplot(x_train['Malic acid'],ax=ax1)

# After Scaling
ax2.set_title('After Scaling')
sns.kdeplot(x_train_scaled['Alcohal'],ax=ax2)
sns.kdeplot(x_train_scaled['Malic acid'],ax=ax2)

plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# Before Scaling
ax1.set_title('Salary distribution Before Scaling')
sns.kdeplot(x_train['Alcohal'],ax=ax1)

# After Scaling
ax2.set_title('Salary distribution After Scaling')
sns.kdeplot(x_train_scaled['Alcohal'],ax=ax2)


plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# Before Scaling
ax1.set_title('Salary distribution Before Scaling')
sns.kdeplot(x_train['Malic acid'],ax=ax1)

# After Scaling
ax2.set_title('Salary distribution After Scaling')
sns.kdeplot(x_train_scaled['Malic acid'],ax=ax2)


plt.show()
```
